# 03 · The survival overview

Establish the baseline before comparing passenger groups. Run notebook 02 first.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
data = pd.read_csv(root / "data" / "cleaned.csv")
plt.rcParams.update({"figure.figsize": (7, 4), "axes.spines.top": False, "axes.spines.right": False})
```

```python
counts = data["Survived"].value_counts().reindex([1, 0], fill_value=0)
summary = pd.DataFrame({"Passengers": counts, "Rate %": (counts / len(data) * 100).round(2)})
summary.index = ["Survived", "Did Not Survive"]
print(summary.to_string())
```

```python
plt.pie(counts, labels=summary.index, colors=["#2ecc71", "#e74c3c"], autopct="%1.2f%%", startangle=90)
plt.title("Survival in the 891-passenger sample")
plt.axis("equal")
plt.show()
```

**Takeaway:** 342 passengers survived (38.38%); 549 did not (61.62%). Survival was the less common outcome in this sample.
